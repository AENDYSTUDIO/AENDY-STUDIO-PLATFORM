### ⚠️ Примечание
Если вы перезагрузили страницу или открыли блокнот заново, обязательно выполните **Шаг 1** и ячейку установки компонентов перед запуском загрузки.

```python
#@title 🚀 Запуск загрузки { display-mode: "form" }

video_url = "https://youtu.be/aEDueh-V-ro?si=_NzhF73bsQ8JzGhv" #@param {type:"string"}
gdrive_path = '/content/drive/My Drive/YoutubeDownloads'

import yt_dlp
import os

def download_to_gdrive(url, output_path):
    ydl_opts = {
        'format': 'bestvideo[ext=mp4]+bestaudio[ext=m4a]/best[ext=mp4]/best',
        'outtmpl': os.path.join(output_path, '%(playlist_title)s/%(title)s.%(ext)s'),
        'noplaylist': False,
        'ignoreerrors': True,
        'quiet': False,
        'no_warnings': True
    }

    try:
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            print(f"🔍 Анализирую: {url}")
            ydl.download([url])
        print(f"\n✨ Готово! Проверьте папку на Диске: {output_path}")
    except Exception as e:
        print(f"❌ Ошибка: {e}")

if os.path.exists('/content/drive/My Drive'):
    if video_url:
        download_to_gdrive(video_url, gdrive_path)
    else:
        print("⚠️ Пожалуйста, введите URL видео.")
else:
    print("🔴 Ошибка: Google Диск не подключен! Запустите Шаг 1.")
```
